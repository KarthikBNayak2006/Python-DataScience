```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import statistics as st
```

## Mean(Avg)

```python
titanicData=pd.read_csv("titanic.csv")
titanicData.head(10)
```

```python
meanValue=np.mean(titanicData["Fare"])
meanValue
```

```python
sns.histplot(x='Fare',data=titanicData,bins=[i for i in range(0,300,20)])
plt.plot([meanValue for i in range(0,300)],[i for i in range(0,300)],c='red')
plt.show()
```

## Median

```python
titanicData=pd.read_csv("titanic.csv")
titanicData.head(10)
```

```python
medianVal=np.median(titanicData["Fare"])
medianVal
```

```python
sns.histplot(x='Fare',data=titanicData,bins=[i for i in range(0,150,10)])
plt.plot([medianVal for i in range(0,300)],[i for i in range(0,300)],c='green')
plt.show()
```

### Final Output

```python
sns.histplot(x='Fare',data=titanicData,bins=[i for i in range(0,150,10)])
plt.plot([meanValue for i in range(0,300)],[i for i in range(0,300)],c='red',label='mean')
plt.plot([medianVal for i in range(0,300)],[i for i in range(0,300)],c='green',label='meadian')
plt.legend()
plt.show()
```

### Mode


```python
modeVal=st.mode(titanicData['Fare'])
modeVal
```

```python
sns.histplot(x='Fare',data=titanicData,bins=[i for i in range(0,150,10)])
plt.plot([modeVal for i in range(0,370)],[i for i in range(0,370)],c='orange')
plt.ylim(0,400)
plt.show()
```

### Final Output

```python
sns.histplot(x='Fare',data=titanicData,bins=[i for i in range(0,150,10)])
plt.plot([meanValue for i in range(0,300)],[i for i in range(0,300)],c='red',label='mean(32.02)')
plt.plot([medianVal for i in range(0,300)],[i for i in range(0,300)],c='green',label='meadian(14.4)')
plt.plot([modeVal for i in range(0,370)],[i for i in range(0,370)],c='orange',label='mode(8.4)')
plt.legend()
plt.show()
```

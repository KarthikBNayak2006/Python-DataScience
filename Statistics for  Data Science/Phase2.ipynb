### MAD(Mean Absolute Deviation)

```python
IT_Department_A = {
    "Employee ID": ["E101", "E102", "E103", "E104", "E105", "E106"],
    "Employee Name": ["Rohan", "Kiran", "Aditya", "Sneha", "Pooja", "Rahul"],
    "Project": [75, 65, 73, 68, 72, 67]
}
IT_Department_B= {
    "Employee ID": ["E201", "E202", "E203", "E204", "E205", "E206"],
    "Employee Name": ["Arjun", "Deepak", "Kavya", "Megha", "Sanjay", "Varun"],
    "Project": [90, 47, 93, 47, 96, 51]
}

deptA_df = pd.DataFrame(IT_Department_A)
deptB_df = pd.DataFrame(IT_Department_B)
```

```python
plt.scatter([i for i in range(0,len(deptA_df["Project"]))],deptA_df["Project"],color='red',s=100,label="Dept A")

plt.scatter([i for i in range(0,len(deptB_df["Project"]))],deptB_df["Project"],color='green',s=100,label="Dept B")
meanVal=deptA_df['Project'].mean()
plt.axhline(y=meanVal,color='orange',linestyle='--',label='Mean')
plt.legend()
plt.show()
```

```python
#Finding MAD for Dept-A
meanVal=deptA_df['Project']
mad_deptA=np.sum(np.abs(deptA_df['Project'] -meanVal))/len(deptA_df['Project'])
mad_deptA

meanVal=deptB_df['Project']
mad_deptB=np.sum(np.abs(deptB_df['Project'] -meanVal))/len(deptB_df['Project'])
mad_deptB
```

### MatPlotLib

```python
import matplotlib.pyplot as plt
import pandas as pd
```

Bar Plot

```python
cname=["Java","Python","C","C++"]
enroll=[24,30,12,14]
plt.bar(cname,enroll)
plt.show()
```

```python
course_data={
    "cname":["Java","Python","C","C++"],
    "enroll":[24,30,12,14]
}
course_df=pd.DataFrame(course_data)
#create bar
plt.Figure(figsize=(9,6))
bars=plt.bar(
    course_df["cname"],
    course_df["enroll"],
    color=["#69E5F2","#EDBC0C","#A326DD","#69F26B"],
    edgecolor="white",
    width=0.5
)
#add title and labels
plt.title("Course Enrollment Overview",fontsize=16,fontweight="bold")
plt.xlabel("Course Name",fontsize=12)
plt.ylabel("Enrollment Students",fontsize=12)
plt.bar_label(bars,padding=3,fontsize=12,fontweight="bold")
plt.ylim(0,55)
plt.gca().spines["top"].set_visible(False)
plt.gca().spines["right"].set_visible(False)  
plt.grid(axis="y",linestyle="--",alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
data=pd.read_excel("expense.xlsx")
exp_df=pd.DataFrame(data)
expgroupby=exp_df.groupby("Category")['Amount'].sum()
plt.bar(expgroupby.index,expgroupby.values, color=["#69E5F2","#EDBC0C","#A326DD","#69F26B"], edgecolor="white")
plt.title("Expense by Category",fontsize=16,fontweight="bold")
plt.xlabel("Category",fontsize=12)
plt.ylabel("Amount",fontsize=12)
plt.ylim(0,5000)
plt.gca().spines["top"].set_visible(False)
plt.gca().spines["right"].set_visible(False)  
plt.grid(axis="y",linestyle="--",alpha=0.5)
plt.tight_layout()
plt.show()
```

Line Plot

```python
weekly_Sales={
    "Day":["Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday"],
    "week1":[15,12,18,8,10,19,30],
    "week2":[25,14,10,18,13,17,35],
}
sales=pd.DataFrame(weekly_Sales)
plt.plot(sales['Day'],sales['week1'],
         marker="d",
         ls="--", label="week1_sales", color="red",
         )
plt.plot(sales['Day'],sales['week2'],
         marker="o",
         ls="-", label="week2_sales", color="blue",
         )
plt.title("Weekly Sales Overview")
plt.xlabel("Day")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

## Scatter plot

```python
#Advertising expenditure and sales data(in ₹ thousands)
adv_sales={
    "Advertising":[100,20,34,30,10,60,10,80,90,50],
    "sales":[130,150,170,190,210,230,250,270,290,310]

}
sales_df=pd.DataFrame(adv_sales)
plt.scatter(sales_df["Advertising"],sales_df["sales"],
            color ="red",
            marker="*",alpha=0.5,
            s=200,label="Sales Data"
            )
plt.title("Advertising Expenditure vs Sales Revenue")
plt.xlabel("Advertising Expenditure (₹ thousands)")
plt.ylabel("Sales Revenue (₹ thousands)")


plt.show()
```

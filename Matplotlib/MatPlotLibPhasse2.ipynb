### Pie Chart

```python
monthly_data={
    "Brand":["onePlus","Apple","Samsung","Vivo","Oppo"],
    "Sales":[16,5,19,23,17]
}
sales_df=pd.DataFrame(monthly_data)
plt.Figure(figsize=(8,6))
plt.pie(sales_df['Sales'],labels=sales_df['Brand'],colors=["#139de2","#21e60c","#650c15",'#ffcc99','#c2c2f0'],
    explode=[0,0,0,0,0.1],
    shadow=True,
    startangle=180 ,
    autopct='%.2f%%',
    pctdistance=0.7,
)  
plt.title('Mobile Sales Distribution',fontsize=12,fontweight='bold')
plt.legend(sales_df['Brand'],title="Mobile Brands", loc='upper right')
plt.tight_layout() 
plt.show()
```

### Stack Plot

```python
Weekly_sales={
     "Day":["Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday"],
        "week1":[15,12,18,8,10,19,30],
        "week2":[25,14,10,18,13,17,35],
        "week3":[10,40,50,23,69,82,43]


}
sales_df=pd.DataFrame(Weekly_sales)
plt.stackplot(
    sales_df['Day'],sales_df['week1'],sales_df['week2'],sales_df['week3'],
    labels=["week1","week2","week3"],
)
plt.legend()
plt.show()
```

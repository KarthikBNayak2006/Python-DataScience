```python
import matplotlib.pyplot as plt
import pandas as pd
```

### Pie Chart

```python
monthly_data={
    "Brand":["onePlus","Apple","Samsung","Vivo","Oppo"],
    "Sales":[16,5,19,23,17]
}
sales_df=pd.DataFrame(monthly_data)
plt.Figure(figsize=(8,6))
plt.pie(sales_df['Sales'],labels=sales_df['Brand'],colors=["#139de2","#21e60c","#650c15",'#ffcc99','#c2c2f0'],
    explode=[0,0,0,0,0.1],
    shadow=True,
    startangle=180 ,
    autopct='%.2f%%',
    pctdistance=0.7,
)  
plt.title('Mobile Sales Distribution',fontsize=12,fontweight='bold')
plt.legend(sales_df['Brand'],title="Mobile Brands", loc='upper right')
plt.tight_layout() 
plt.show()
```

```python
#pie chart for expense with upi,cash and card
data=pd.read_excel("expense.xlsx")
exp_df=pd.DataFrame(data)
plt.figure(figsize=(8,6))
plt.pie(exp_df['Amount'],labels=exp_df['Category'],colors=["#139de2","#21e60c","#650c15",'#ffcc99','#c2c2f0'],
    explode=[0,0,0,0,0.1],
    shadow=True,
    startangle=180 ,
    autopct='%.2f%%',
    pctdistance=0.7,
)  
plt.title('Expense Distribution',fontsize=12,fontweight='bold')
plt.legend(exp_df['Category'],title="Expense Categories", loc='upper right')
plt.tight_layout() 
plt.show()
```

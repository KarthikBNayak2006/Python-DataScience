Numpy Function


```python
import numpy as np
```

### Aggregate Function

```python
arr=np.array([10,20,30,40,50])
print("Sum:",np.sum(arr))
print("Max:",np.max(arr))
print("Min:",np.min(arr))
print("Count:",np.size(arr))
print("Cumulative Sum:",np.cumsum(arr))
print("Cumulative product",np.cumprod(arr))
```

### Statistical Function

```python
#suger 20gm
import statistics as st #used for mode
m1=np.array([18,22,17,25,20,18,23])
m2=np.array([20,15,25,30,10,50,35])
print("Mean M1:",np.mean(m1),"Mean M2:",np.mean(m2))
print("Median M1:",np.median(m1),"Median M2:",np.median(m2))
print("Mode M1:",st.mode(m1),"Mode M2:",st.mode(m2))
print("Standerd Division M1:",np.std(m1),"Standerd Division M2:",np.std(m2))
```

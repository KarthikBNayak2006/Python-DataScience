## Student Attendance Analytics with Pandas

```python
import pandas as pd
import numpy as np
```

#### Section 0: Data Loading

```python
df = pd.read_excel("csea_attendance.xlsx")
df
```

#### Section 1: Data Inspection

```python
"""Q1. Display the first 10 rows, the shape of the DataFrame, and the data types of each column."""
print(df.head(10))
print('Shape:', df.shape)
print('Data types:')
print(df.dtypes)
```

```python
"""Q2. Check for missing values in each column. Identify which columns have null values (especially Last Present Date with "Never") and report the count. """
df['Last Present Date'] = df['Last Present Date'].replace('Never', pd.NaT)
missing_counts = df.isna().sum()
print(missing_counts)
print('Columns with missing values:')
print(missing_counts[missing_counts > 0])
```

```python
"""Q3. Rename the columns to snake_case (e.g., Roll No → roll_no, Student Name → student_name, Attendance % → attendance_pct). Display the updated column list."""
df = df.rename(columns={'Roll No': 'roll_no', 'Student Name': 'student_name', 'Subject': 'subject', 'Total Classes': 'total_classes', 'Present': 'present', 'Absent': 'absent', 'Excused': 'excused', 'Attendance %': 'attendance_pct', 'Status': 'status', 'Classes Needed (75%)': 'classes_needed_75', 'Last Present Date': 'last_present_date', 'Consecutive Absences': 'consecutive_absences'})
print(df.columns.tolist())
```

```python
"""Q4. Convert Attendance % from string format (e.g., "64%") to a numeric float column (0.64 or 64.0). Also convert Last Present Date to datetime, handling "Never" as NaT."""
df['attendance_pct'] = pd.to_numeric(df['attendance_pct'].astype(str).str.rstrip('%'), errors='coerce')
df['last_present_date'] = pd.to_datetime(df['last_present_date'], errors='coerce')
print(df[['attendance_pct', 'last_present_date']].head())
```

#### Section 2: Filtering & Selection

```python
"""Q5. Filter and display all students whose Status is "Critical". Sort them by Attendance % in ascending order."""
critical_students = df.loc[df['status'] == 'Critical'].sort_values('attendance_pct')
print(critical_students)
```

```python
"""Q6. Extract all students who have 0 present days (fully absent). List their roll numbers, names, and consecutive absences."""
fully_absent = df.loc[df['present'] == 0, ['roll_no', 'student_name', 'consecutive_absences']]
print(fully_absent)
```

```python
"""Q7. Using .loc and .iloc, display:
(a) All "Safe" students using .loc
(b) The first 15 rows and columns 2-6 using .iloc """
print('Safe students (.loc):')
print(df.loc[df['status'] == 'Safe'])
print('First 15 rows, columns 2-6 (.iloc):')
print(df.iloc[:15, 2:7])
```

```python
"""Q8. Find all students whose Consecutive Absences is greater than or equal to 5. Display their name, status, and last present date."""
high_absence = df.loc[df['consecutive_absences'] >= 5, ['student_name', 'status', 'last_present_date']]
print(high_absence)
```

### Section 3: Grouping & Aggregation 

```python
"""Q9. Group the data by Status and compute: 
a) Count of students 
b) Average attendance %  
c) Maximum consecutive absences"""
status_summary = df.groupby('status').agg(student_count=('student_name', 'count'), average_attendance_pct=('attendance_pct', 'mean'), max_consecutive_absences=('consecutive_absences', 'max'))
print(status_summary)
```

```python
"""Q10. Create a new column Attendance Category:
≥ 90% → "Excellent"
75-89% → "Good"
60-74% → "Average"
< 60% → "Poor"
Then use groupby() to count students in each category."""
df['attendance_category'] = pd.cut(df['attendance_pct'], bins=[-float('inf'), 60, 75, 90, float('inf')], labels=['Poor', 'Average', 'Good', 'Excellent'], right=False)
print(df.groupby('attendance_category', observed=False).size())
```

```python
"""Q11. Calculate the total number of classes conducted (sum of Total Classes) and the total present/absent counts across all students using .agg()."""
totals = df[['total_classes', 'present', 'absent']].agg('sum')
print(totals)
```

```python
"""Q12. Group by Status and use .agg() to find:
a)min, max, and mean of Present
b)sum of Absent"""
status_attendance = df.groupby('status').agg(present_min=('present', 'min'), present_max=('present', 'max'), present_mean=('present', 'mean'), absent_sum=('absent', 'sum'))
print(status_attendance)
```

### Section 4: Feature Engineering & Transformation

```python
"""Q13. Create a new column Risk Score using the formula:
Risk Score = (Absent * 0.4) + (Consecutive Absences * 0.6)
Display the top 10 students with the highest risk score."""
df['risk_score'] = (df['absent'] * 0.4) + (df['consecutive_absences'] * 0.6)
print(df.nlargest(10, 'risk_score'))
```

```python
"""Q14. Add a column Classes Needed (Numeric) by extracting the numeric value from the Classes Needed (75%) column (e.g., "5 more" → 5, "✓ Met" → 0)."""
df['classes_needed_numeric'] = pd.to_numeric(df['classes_needed_75'].astype(str).str.extract(r'(\d+)')[0], errors='coerce').fillna(0).astype(int)
print(df[['classes_needed_75', 'classes_needed_numeric']].head())
```

```python
"""Q15. Using apply() with a lambda function, create a column Attendance Flag that returns:
a)"Defaulter" if attendance < 75%
b)"Safe" otherwise """
df['attendance_flag'] = df['attendance_pct'].apply(lambda attendance: 'Defaulter' if attendance < 75 else 'Safe')
print(df[['student_name', 'attendance_pct', 'attendance_flag']].head())
```

```python
"""Q16. Use pd.cut() to bin students into attendance ranges: [0-50, 50-75, 75-90, 90-100] and label them as ["Critical", "Warning", "Good", "Excellent"]."""
df['attendance_band'] = pd.cut(df['attendance_pct'], bins=[0, 50, 75, 90, 100], labels=['Critical', 'Warning', 'Good', 'Excellent'], include_lowest=True)
print(df[['student_name', 'attendance_pct', 'attendance_band']].head())
```

### Section 5: Advanced Analysis 

```python
"""Q17. Using pivot_table(), create a summary showing the count of students by Status and Attendance Category (created in Q10)."""
status_category_counts = pd.pivot_table(df, index='status', columns='attendance_category', values='student_name', aggfunc='count', fill_value=0, observed=False)
print(status_category_counts)
```

```python
"""Q18. Identify duplicate student names (e.g., "ANANYA", "DISHA") and display all rows where the base name repeats. Use string manipulation to extract the base name."""
df['base_name'] = df['student_name'].str.upper().str.replace(r'\([^)]*\)', '', regex=True).str.strip().str.split().str[0]
name_counts = df['base_name'].value_counts()
repeated_base_names = name_counts[name_counts > 1].index
repeated_rows = df.loc[df['base_name'].isin(repeated_base_names)].sort_values(['base_name', 'student_name'])
print(repeated_rows)
```

```python
"""Q19. Merge this dataset with a hypothetical Student_Details DataFrame containing Roll No, Branch, and Semester. Perform an inner join and display the merged result."""
student_details = pd.DataFrame({'roll_no': df['roll_no'], 'branch': 'CSE', 'semester': 3})
merged_df = df.merge(student_details, on='roll_no', how='inner')
print(merged_df)
```

```python
"""Q20. Export the following to separate CSV files:
(a) All "Critical" students
(b) Summary statistics grouped by Status
(c) Top 10 students by Risk Score"""
critical_students = df.loc[df['status'] == 'Critical']
status_summary = df.groupby('status').agg(student_count=('student_name', 'count'), average_attendance_pct=('attendance_pct', 'mean'), max_consecutive_absences=('consecutive_absences', 'max'))
top_10_risk_students = df.nlargest(10, 'risk_score')
critical_students.to_csv('critical_students.csv', index=False)
status_summary.to_csv('status_summary.csv')
top_10_risk_students.to_csv('top_10_risk_score.csv', index=False)
print('Exported critical_students.csv, status_summary.csv, and top_10_risk_score.csv')
```
